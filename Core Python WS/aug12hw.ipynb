## Q1. Write a Python program to input a number and check whether it is a two-digit number or not.

```python
num = int(input("Enter a number:"))

if 10 <= num <= 99:
    print(num,"is a Two digit number")
else:
    print(num,"is not a two digit number")
```

## Q2. Write a Python program to input three numbers and find the largest number.

```python
num1 = int(input("Enter 1st number:"))
num2 = int(input("Enter 2nd number:"))
num3 = int(input("Enter 3rd number:"))

if num1 > num2 and num1 > num3:
    print("1st number:",num1)
    print("2nd number:",num2)
    print("3rd number:",num3)
    print(num1,"is Largest number")
elif  num2 > num1 and num2 > num3:
    print("1st number:",num1)
    print("2nd number:",num2)
    print("3rd number:",num3)
    print(num2,"is Largest number")
elif  num3 > num1 and num3 > num2:
    print("1st number:",num1)
    print("2nd number:",num2)
    print("3rd number:",num3)
    print(num3,"is Largest number")
else:
    print("1st number:",num1)
    print("2nd number:",num2)
    print("3rd number:",num3)
    print("All numbers are equal")
```

## Q3. Write a Python program to input three sides of a triangle and check whether the sides can form a valid triangle.

```python
side1 = int(input("Enter 1st side: "))
side2 = int(input("Enter 2nd side: "))
side3 = int(input("Enter 3rd side: "))

if side1 + side2 >= side3 and side2 + side3 >= side1 and side1 + side3 >= side2:
    print("1st Side:",side1)
    print("2nd Side:",side2)
    print("3rd Side:",side3)
    print("Valid Triangle")
else:
    print("1st Side:",side1)
    print("2nd Side:",side2)
    print("3rd Side:",side3)
    print("Invalid Triangle")
```
