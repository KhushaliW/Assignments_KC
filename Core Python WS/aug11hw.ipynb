# Composite assignment

A composite assignment (more commonly called a compound assignment) in programming is a short way to perform a math or logical operation on a variable and assign the new result back to that same variable in one single step.

### += , -= , *= , /= , //= , %= , **=

```python
# 
a = 15
a +=5
print(a) # 20

b = 15
b -=5
print(b) # 10

c = 15
c *=5
print(c) # 75

d = 15
d /=5
print(d) # 3.0

e = 15
e //=5
print(e) # 3

f = 15
f %=5
print(f) # 0

g = 15
g **=5
print(g) # 3375
```

# INTEGERS
### 1.Decimal 
 0 to 9 Base 10 system

```python
var=10
print(var) # 10
print(type(var)) # <class 'int'>
```

### 2.Binary
 0 and 1 Base 2 system

```python
var=0b1011
print(var) # 11
print(type(var)) # <class 'int'>
```

### 3.Octal

```python
var=0o10
print(var) # 529
print(type(var)) # <class 'int'>
```

### 3.Hexadecimal

# Bitwise Oprators

Bitwise operators in Python are used to perform operations directly on the binary digits (bits) of integers.

### & , | , ^ , ~ , >> , << 


```python
# Bitwise AND
a = 2 & 3
print(a) # 2

# Bitwise OR
a = 2 | 3
print(a) # 3

# Bitwise XOR
a = 2 ^ 3
print(a) # 1

# Bitwise NOT
a = ~ 1
print(a) # -2

# Bitwise Left Shift
a = 5 << 2
print(a) # 20

# Bitwise Right Shift
a = 5 >> 2
print(a) # 1
```
