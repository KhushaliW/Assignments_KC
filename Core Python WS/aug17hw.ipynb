# String methods
-> built-in functions used to manipulate and work with strings.

## 1. find() - Find the position of text


```python
text1 = "This is find() method"
print(text1.find("method"))
```

## 2. strip() — Remove spaces from both sides

```python
text2 = "     Remove    "
print(text2)
print(text2.strip())
print(text2.rstrip())
print(text2.lstrip())
```

## 3. swapcase() — Swap uppercase and lowercase

```python
text3 = "kHUsHaLi"
print(text3.swapcase())
```

## 4. replace() — Replace text

```python
text4 = "This is swapcase() method"
print(text4.replace("swapcase()", "replace()"))
```

## 5. startswith() — Check beginning

```python
text5 = "This is startswith() method"
print(text5.startswith("This"))
print(text5.startswith("method"))
```

## 6. endswith() — Check ending

```python
text6 = "This is endswith() method"
print(text6.endswith("This"))
print(text6.endswith("method"))
```

## 7. isupper() — Check whether all letters are uppercase

```python
text7 = "KHUSHALI"
print(text7.isupper())

subtext7 = text7.lower()
print(subtext7.isupper())
```

## 8. islower() — Check whether all letters are lowercase

```python
text8 = "khushali"
print(text8.islower())

subtext8 = text8.upper()
print(subtext8.islower())
```

## 9. isspace() — Check whether string contains only spaces

```python
text9 = "       "
print(text9.isspace())

subtext9 = text9.strip()
print(subtext9.isspace())
```

## 10. split() — Convert string into a list

```python
text10 = "This is split() method"
print(text10.split())
```
