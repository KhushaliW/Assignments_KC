# Python List Methods

| ID | Method | Description | 
|----|--------|-------------|
| 1 | `copy()` | Creates a copy of the list |
| 2 | `insert()` | Adds an element at a specific index |
| 3 | `extend()` | Adds multiple elements to the list |
| 4 | `remove()` | Removes a specific element | 
| 5 | `pop()` | Removes an element using its index |
| 6 | `clear()` | Removes all elements from the list |
| 7 | `index()` | Returns the index of an element | 
| 8 | `count()` | Counts how many times an element occurs |
| 9 | `sort()` | Sorts the list |
| 10 | `reverse()` | Reverses the list |
| 11 | `append()` | Adds an element at the end of the list | 

## 1. copy() — Create a copy of a list

```python
rollNo = [101, "Ruhi", 104, 105, "Riya", "Niti", 108, 109]

new_rollNo = rollNo.copy()

print("Original Roll No:",rollNo)
print("Copy Roll No:",new_rollNo)
```

## insert() — Add an element at a specific position
### Syntax: list.insert(index, value)

```python
rollNo.insert(1, 102)

print("List After Insertion:",rollNo)
```

## 3. extend() — Add multiple elements

```python
rollNo.extend([110, 110])

print("List After Extending:",rollNo)
```

## 4. remove() — Remove a specific value

```python
rollNo.remove(102)

print("List After Removing:",rollNo)
```

## 5. pop() — Remove an element using its index

```python
rollNo.pop(4)

print("List After Removing:",rollNo)
```

## 6. clear() — Remove all elements

```python
print("List Before Clearing:",new_rollNo)
new_rollNo.clear()

print("List After Clearing:",new_rollNo)
```

## 7. index() — Find the index of an element

```python
print("Index of 105:",rollNo.index(105))
```

## 8. count() — Count how many times a value appears

```python
print("How many times 105 is present:",rollNo.count(110))
```

## 9. sort() — Sort the list

```python
new_rollNo = [105, 103, 101, 106, 102, 104, 99]

print("Original Roll No:",new_rollNo)
print("Sorted Roll No:",sorted(new_rollNo))

print("Original Roll No:",new_rollNo)
new_rollNo.sort(reverse=True)
print("Reverse Sorted Roll No:",new_rollNo)
```

## 10. reverse() — Reverse the list

```python
print("Original Roll No:",new_rollNo)
new_rollNo.reverse()
print("Reversed Roll No:",new_rollNo)
```

## 11. append() - Adds an element at the end of the list

```python
print("List Before Appending:",rollNo)
rollNo.append(111)
print("List After Appending:",rollNo)
```
